# Operators
## 1) Arithematic operator
- '+' : Addition (str-> Concatenation)
- '-' : Substraction
- '*' : Multiplication (str-> Repetion)
- '/' : Division
- '%' : Modulo
- '//' : Floor Division
- '**' : Exponent or Power

```python
a=12
b=5
print("a + b: ",a+b)
print("a - b: ",a-b)
print("a * b: ",a*b)
print("a / b: ",a/b)
print("a % b: ",a%b)
```

## 2) Comparision operators
- '>' : Greater than
- '<' : Less than
- '<=' : Greater than or equal to
- '>=' : Less than or equal to
- '==' : Equal to
- '!=' : Not equal to

## 3) Logical operators
### - For boolean type behaviour
- AND : If both arguments are true then only result is true
- OR : If atleast one argument is true then result is true
- NOT : Complement
### - For non-boolean type behaviour
- X AND Y : If x evaluates to false then return x otherwise return y
- X OR Y : If x evaluates to true then return x otherwise return y
- NOT X : If x evaluates to false then result is true otherwise result is false

## 4) Ternary or conditional operators
### x=first value if condition else second value

```python
a=10
b=20
c=30 if a>b else 40
print(c)
```

## 5) Assignment operators
- '=' : x=10
- '+=' : x+=10 -> x=x+10
- '-=' : x-=10 -> x=x-10
- '\*=' : x\*=10 -> x=x*10
- '/=' : x/=10 -> x=x/10
- '//=' : x//=10 -> x=x//10
- '%=' : x%=10 -> x=x%10
- '**=' : x\**=10 -> x=x/**10
## 6) Membership operators
- IN : x="Python is easy"
        print("i" in x)
- NOT IN : x="Python is easy"
        print("i" not in x)
