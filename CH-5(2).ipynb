```python
# list comprehensions.
# Synatx:- list=[expession for item in list if condition]
s=[x*x for x in range(1,11)]
print(s)
v=[x for x in s if x%2==0]
print(v)
z=[2**x for x in range(1,6) if x%2==0]
print(z)
```

```python
# wap to perform a circular shift on a list to the right direction shift is to be entered by user and if user enters 7 then reverse the list.
l=[1,2,3,4,5,6,7]
n=int(input('Enter your shift'))
if n==7:
    l.reverse()
    print(l)
else:
    x=l[-n:]+l[:-n]
    print(x)
```
