### keys()

### WAP that reads the words from paragraph and stores them as keys in a dictionary and count the frequency of it as a value

```python
s="Dog the quick brown for jumps over the dog"
words=s.split()
print(words)
dict={}
for i in words:
    dict[i]=words.count(i)
print(dict)
```

### WAP to create a dictionary with the keys as first character and the values as a list of words starting with that character

```python
s="Don't wait for your feeling to change to take action. Take the action and your feelings will change"
words=s.split()
dict={}
for i in words:
    first=i[0]
    if first in dict:
        dict[first].append(i)
    else:
        dict[first]=[i]
print(dict)
```


### WAP to replace words from dictionary. You are given a string, replace its words from given dictionary

```python
s="Don't wait for your feeling to change to take action. Take the action and your feelings will change"
d={"to":"go","change":"no change","action":"lazy"}
for i,j in d.items():
    s=s.replace(i,j)
print(s)
```

### OR

```python
s="Don't wait for your feeling to change to take action. Take the action and your feelings will change"
d={"to":"go","change":"no change","action":"lazy"}
l=[]
for i in s.split():
    if i in d:
        l.append(d[i])
    else:
        l.append(i)
x=" ".join(l)
print(x)
```

### union()

```python
x={10,20,30,}
y={30,40,50,60}
print(x.union(y))  #( x|y )
print(x)
print(y)
```

### intersection()

```python
x={10,20,30,}
y={30,40,50,60}
print(x.intersection(y))  #(x&y)
print(y.intersection(x))  #(y&x)
print(x)
print(y)
```

### difference()

```python
x={10,20,30,}
y={30,40,50,60}
print(x.difference(y))  #(x-y)
print(y.difference(x))  #(y-x)
print(x)
print(y)
```

### symmetric_difference()

```python
x={10,20,30,}
y={30,40,50,60}
print(x.symmetric_difference(y))  #( x^y )
print(y.symmetric_difference(x))  #( y^x )
print(x)
print(y)
```

### issubset()

```python
x={"a","b","c"}
y={"d","e","c","b","a"}
print(x.issubset(y)) 
print(y.issubset(x))  
print(x)
print(y)
```

### issuperset()

```python
x={"a","b","c"}
y={"d","e","c","b","a"}
print(x.issuperset(y)) 
print(y.issuperset(x))  
print(x)
print(y)
```
