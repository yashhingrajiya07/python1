### Accessing Elements Of List.
1. using index
2. using slicing operation

```python
l=[10,20,30,40,50]
print(l[0])
print(l[-1])
#print(l[10]) generates an error.
```

```python
l=list(range(1,11))
print(l)
print(l[2:7:2])
print(l[4::2])
print(l[3:7])
print(l[8:2:-2])
print(l[8:2:2])
print(l[-2:-8:-2])
print(l[4:100])
print(l[::-1])
```

### IMP Functions Of List

```python
# 1. len()
l=[1,2,3,4,5]
print(len(l))
```

```python
l=[10,20,30,40,50]
y=l[:] # if we modify something in y then l(original) also changes so we can use slicing operator if we don't want any change in the original one.
y=l.copy() # also we can use copy function then also the original list will remain as it is.
y[0]=100
print(l)
print(y)
print(id(l))
print(id(y))
```

```python
# 2. count()
l=[10,20,30,10,40,50,10]
print(l.count(10))
```

```python
# 3. index()
n=[1,2,3,3,3,1]
print(n.index(1))
print(n.index(2))
#print(n.index(4)) generates an error.
if 4 in n:
    print(n.index(4))
```

```python
# 4. append()
l=[10,20,30,40,50]
l.append(60)
l.append('demo')
print(l)
```

```python
l=[]
for i in range(5):
    x=int(input('Enter Number'))
    l.append(x)
print(l)
```

```python
# 5. insert() -> uses to enter the element in between or any particular index.
l=[1,2,4,5]
l.insert(2,3) # first argument we have to provide index and the 2nd one is what do we want to insert.
l.insert(100,6) # if we give the range out of length then it is entered at the last one.
l.insert(-8,0) # it is inserted at first.
print(l)
```

```python
# 6. extend()
l1=['apple','mango']
l2=['orange','cherry']
l1.extend(l2)
print(l1)
l2.extend(l1)
print(l2) # because l1 is updated so length becomes 6.
```

```python
l=['Arman','Dhairya']
l.extend('Aryan') # the string becomes list but each char is seperated.
print(l)
```

```python
# 7. remove()
n=[1,2,3,1,5,4]
n.remove(1)
print(n)
```

```python
n=[1,2,3,4,5]
n.insert(-1,10)
print(n)
```

```python
# 8. pop()
n=[10,20,30,40]
print(n.pop())
print(n.pop())
print(n.pop())
print(n)
```

```python
n=[10,20,30,40]
n.pop(1) # pops the element at 1st index.
print(n)
```

```python
y=[]
print(y.pop(2)) # this gives index error.
print(y.remove(2)) # this gives value error.
```

```python
# 9. clear()
n=[1,2,3,4,5]
n.clear()
print(n)
```

```python
# 10. reverse()
n=[1,2,3,4,5]
n.reverse()
print(n)
```

```python
# 11. sort()
n=[2,5,1,10]
n.sort()
print(n)
n.sort(reverse=True) # sorted then given in the reverse order so in descending order we can get.
print(n)
n=['D','B','C','A']
n.sort()
print(n)
```

```python
a=['banana','apple','strawberry','orange']
a.sort()
print(a)
a.sort(reverse=True)
print(a)
```

```python
a.sort(key=len,reverse=True) # here we have given key which means that it is sorted acc to length.
print(a)
```

```python
a=[(1,3),(2,2),(3,1)]
def fun(val):
    return(val[1])
a.sort(key=fun)
print(a)
```
