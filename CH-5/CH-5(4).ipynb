- Set Comprehension:

```python
s1={x+x for x in range(1,11)} #set
print(s1)
s2={x:x+x for x in range(1,11)} #dictionary
print(s2)
```

```python
s=["abc@gmail.com","abcxyz@yahoo.com","xyz@facebook.com","pqr@gmail.com","pqr@yahoo.com"]
a=set()
for i in s:
    b=i.split("@")
    a.add(b[1])
print(a)
```

# LAMBDA FUNCTION

```python
def square(n):
    return n*n
square(2)
```

## MAP()
- for every element present in the given sequence, apply some functionalty and generate new element with required modification
- syntex:- map(function,sequence)

```python
l=[1,2,3,4,5]
def double(x):
    return 2*x
l1=list(map(double, l))
print(l1)
```

```python
l=[1,2,3,4,5]
l1=list(map(lambda x:2*x,l))
print(l1)
```

```python
l1=[1,2,3,4,5,]
l2=[4,5,6,7,8,]
l=list(map(lambda x,y:x+y,l1,l2))
print(l)
```

## REDUCE():
- reduce the sequence of elements into a single element by applying specified function.
- syntex:- reduce(function,sequence)
- reduce function present in functools module so we need to import the module to use it.

```python
from functools import *
l=[10,20,30,40,50,]
sum=reduce(lambda x,y:x+y,l)
print(sum)
mul=reduce(lambda x,y:x*y,l)
print(mul)
```

## filter():
- we can use filter function to filter values from given sequence based on some condition
- syntex:- filter(function , sequence)

```python
def iseven(x):
    if x%2==0:
        return True
    else:
        return False
l=[5,10,15,20,30,35,40,50]
l1=list(filter(iseven,l))
print(l1)
```

```python
l=[5,10,15,20,25,30,40,45]
l1=list(filter(lambda x:x%2==1,l))
print(l1)
```

## frozenset():
- immutable version of set

```python
s=["apple","banana","cherry"]
f=frozenset(s)
print(f)
print(type(f))
f.add("mango")
print(f)
```

```python
x=set(s)
print(x)
print(type(x))
x.add("mango")
print(x)
```
