# Unit:-02 Conditional Execution And Iteration

## Simple if
Syntax:-
if condition:
    Statement-1
    Statement-2
    Statement-3

## If-Else
Syntax:-
if condition:
    Statement-1
else 
    Statement-2

```python
a=input("Enter Name:")
if a=="Arman":
    print("Hello Arman")
else:
    print("Hello Guest")
    print('How are you?')
```

## If-elif-else
Syntax:-
if condition1:
    action-1
elif condition2:
    action-2
elif condition3:
    action-3
else:
    default action

```python
n1=int(input("Enter n1:"))
n2=int(input("Enter n2:"))
n3=int(input("Enter n3:"))
if n1>n2 and n1>n3:
    print("Biggest Number:",n1)
elif n2>n3:
    print("Biggest Number:",n2)
else:
    print("Biggest Number:",n3)
```

```python
# Nested if
x=41
if x>10:
    print("Above 10")
if x>20:
    print("And also above 20")
else:
    print("But not above 20")
```
