- FIND() -> returns the index first occurence of the given substring if it is not available the we will get -1

```python
s="learning python is very easy"
print(s.find("python"))
print(s.find("Python"))
print(s.find("e"))
print(s.find("p",2,15)) # it dinfs between index 2 & 15
print(s.find("h",2,10))
```

- COUNT() -> returns the number of occurence od the given substring in the string and returns 0 id couldn't find

```python
s="sdfcgh dxfgh"
print(s.find(" "))
print(s.count("x"))
print(s.count("x",0,6))
```

- REPLACE(old string ,new string) -> replace old string with new string

```python
s="learning java is very helpful"
s1=s.replace("java","python")
print(s1)
```

- TRANSLATE() WITH NALETRANS FUNCTION

```python
import string
s="py$t@hon is very ea$sy"
print(string.punctuation)
x=s.maketrans("","",string.punctuation)
y=s.maketrans("","","$") # replace $ in None
z=s.maketrans("e","w","$") # replaxe e in w and $ in None
z=s.maketrans("aes","wpq","$") # replaxe a in w and e in p and s in q and $ in None
#length of both parameters must be same
print(x)
print(y)
print(z)
print(len(x))
print(len(y))
print(len(z))
print(s.translate(x))
print(s.translate(y))
print(s.translate(z))
```

```python
# WAP to display all the position of a substring in a given string
s1=input("enter string")
s2=input("enter substring")
flag=False
pos=-1
n=len(s1)
c=0
while True:
    pos=s1.find(s2,pos+1,n)
    if pos==-1:
        break
    c=c+1
    print("position:",pos)
    flag=True
if flag==False:
    print("not found")
print("No. of occurence:",c)
```

```python
#WAP to merge the characters of 2 string into a single string by taking character's alter value
s1=input("enter string")
s2=input("enter string")
output=""
i,j=0,0
while i<len(s1) or j<len(s2):
    output+=s1[i]
    i+=1
    output+=s2[j]
    j+=1
print(output)
```

- SORTED() -> it sort the string

```python
x="python"
d="".join(sorted(x))
print(d)
```

```python
s="B4A1D3"
s1=""
s2=""
for i in s:
    if i.isalpha():
        s1+=i
    if i.isdigit():
        s2+=i
s1="".join(sorted(s1))
s2="".join(sorted(s2))
output=s1+s2
print(output)
```
