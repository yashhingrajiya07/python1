- ### Enumerate(iterable,start=0)
        iterates the string or tupple and gives individual numbers to all individual entities

```python
l1=("Apple","Banana","Cherry")
s="YoOo"
obj1=list(enumerate(l1))
print(obj1)
obj2=list(enumerate(l1,start=1))
print(obj2)
obj3=list(enumerate(s))
print(obj3)
obj4=enumerate(l1)
obj5=enumerate(s)
print(obj4)
print(obj5)
for i,j in enumerate(l1):
    print(i,j)
```

```python
t=(12,32,5,65,78)
max=t[0]
min=t[0]
for i,j in enumerate(t):
    if j>max:
        max=j
    if j<min:
        min=j
print(max)
print(min)
```

```python
t=(10,15,30,40,42,73,85,88,32)
sumev=0
sumodd=0
for i,j in enumerate(t):
    if j%2==0:
        sumev+=j
    else:
        sumodd+=j
print(sumev)
print(sumodd)
```
