## Remarks:
- False=0
- True=1
- whenever there is "/", the answer will always be in float

```python
print(5**3**1*5/True)
```

```python
print(5**3**False)
```

```python
a=3
print("5" if a>2 else "8") #Ternary Operator
```

```python
print(5 and False or 3/0)
```

```python
print(5 and True or 5/0) #Here it gets true in 5 and True and since there is or after it, it automatically gives true without checking 
```

```python
print("1" in "123" and "False" or True) #The expression prints False because of how Python handles and and or logic:
                                        #1. "1" in "123" is True.
                                        #2. True and "False" stops at the second value, giving you the string "False".
                                        #3. "False" or True looks at "False". Since it is a filled string, Python sees it as a "true" value and immediately stops, returning "False".
                                        #Finally, print() outputs that final string result: False.
```

### Write a program to convert farenheit to celsuis and vica versa

```python
far=float(input("Enter value of temperature in farenheit: "))
ans1=(far-32)*(5/9)
print("Value of ",far,"F is ",ans1," degrees celsuis")
cel=float(input("Enter value of temperature in Celcius: "))
ans2=(cel*(9/5))+32
print("Value of ",cel,"degrees celcius is ",ans2,"F")
```

### Convert total days in years months and days format

```python
totaldays=int(input("Enter total number of days: "))
years=totaldays//365
months=(totaldays % 365) //30
x=(years+1 if months==12 else years)
y=(0 if months==12 else months)
days=(totaldays % 365) % 30
print(x," Year(s), ",y," Month(s), ",days," Day(s)")
```

### Convert total money into number of notes

```python
money=int(input("Enter the amount of total money: "))
notes2000=money//2000
notes500=(money%2000)//500
notes200=((money%2000)%500)//200
notes100=(((money%2000)%500)%200)//100
notes50=((((money%2000)%500)%200)%100)//50
notes20=(((((money%2000)%500)%200)%100)%50)//20
notes10=((((((money%2000)%500)%200)%100)%50)%20)//10
change=((((((money%2000)%500)%200)%100)%50)%20)%10
print("₹2000 notes: ",notes2000)
print("₹500 notes : ",notes500)
print("₹200 notes : ",notes200)
print("₹100 notes : ",notes100)
print("₹50 notes  : ",notes50)
print("₹20 notes  : ",notes20)
print("₹10 notes  : ",notes10)
print("Change     : ",change)
```
