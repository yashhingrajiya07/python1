## writing the files

- write() - write the content in the file without adding any characters
- writeline() - write the content of a list into file

```python
with open("demo.txt",'w') as file:
    file.write("this is demo file\n")
    file.write("this is demo text\n")
```

```python
l=["this is demo file\n","this is demo text\n","lj university"]
f=open("writeline.txt","w+")
f.writelines(l)
f.seek(0)
print(f.read())
f.close()
```

```python
# WAP TO COPY CONTENT ON ONE FILE TO ANOTHER FILE BUT LINE WISE REVERSE
```

```python
f=open("copy.txt")
x=f.readlines()
f1=open('recerse.txt','w')
for i in x[::-1]:
    if i.endswith("\n"):
        f1.write(i)
    else:
        f1.write(i+"\n")
f.close()
f1.close()
```

```python
f=open('hashfile.txt')
f1=open('output.txt','w')
while True:
    b=f.readline()
    if len(b)!=0:
        if b[0]=="#":
            continue
        else:
            if "#" in b:
                for index in range(1,len(b)):
                    if b[index]=="#":
                        f1.write(b[0:index]+"\n")
            else:
                f1.write(b)
    else:
        break
f.close()
f1.close()
```
