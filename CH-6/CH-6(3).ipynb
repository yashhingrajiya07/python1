```python
#WAP to search for a string from the text file. If the string is found then display the position of that string or else display string not found
f=open("copy.txt")
s=input("Enter string that you want to search for:")
a=f.readlines()
pos=0
line=0
flag=True
for i in range(len(a)):
    if a[i].find(s)!=-1:
        pos=a[i].find(s)
        flag=False
        line=i
        break
    else:
        continue
if flag:
    print("String not found")
else:
    print("Line:",line+1)
    print("Index:",pos)
```

```python
#WAP to read a text file and reverse the words starting with i
f1=open("Demo.txt")
a=f1.readlines()
f1.close()

with open("Demo.txt","w+") as f2:
    for i in a:
        b=i.split()
        for j in range(len(b)):
            if b[j].startswith("i") or b[j].startswith("I"):
                w=b[j]
                b[j]=w[::-1]
        x=" ".join(b)+"\n"
        print(x)
        f2.write(x)
```
