## Working With Files
- Text Files: Text terminated with a special character '\n'
- Binary Files: Machine understandable binary language  
### File Access Modes
- Read Mode (r)
- Read and Write (r+)
- Write (w)
- Write and Read (w+) (whenever we open file in w+ mode, all the data is deleted and the data gets overridden meanwhile in R+ mode we can read and write)  
- Append (a) (append always adds data at the end of the file)  
- Append and Read (a+)  

```python
f=open('Text.txt','r') #If we dont give any mode then by default it will be considered read mode
print(f.read())
f.close()
f.closed
```

```python
f=open('Text.txt','r')
f.read() #'Welcome to python\nHi\nHellow \nHow are you\nHola, Como estas'
f.closed
```

```python
with open('Text.txt','r') as f:print(f.read())
f.closed
```

```python
f=open('Demo.txt') #If this is done in read mode then it will give error, but if it is opened in write mode then it will create a new file of such name
```

```python
f.write("Hi") #Can't write in a read only file
```

```python
f=open('utf8_demo.txt')
f.read() #Gives error because the data in it is written in different languages or fonts
```

```python
f=open('utf8_demo.txt','r',encoding='utf8') #Encoding is used for characters that are not supported
print(f.read())
```

```python
f=open('utf8_demo.txt','r',encoding='utf8')
f.read()
```

```python
f=open('Text.txt','r+')
print(f.read()) #cursor goes at the end after we do f.read but if we remove this line, all the stuff starts getting written from the start
f.write('New Data\n')
f.write('My Data')
f.close()
```

```python
f=open('Text.txt','w') #In write only mode or write and read mode it overrides the whole data and file goes blank the moment we open it on 'w' mode
f.write('My Data')
f.close()
```

- f.seek(index): it moves cursor to given index
- f.read(index): reads the amount of index provided after the initial position of cursor
- f.tell(): it tells the position of cursor
- w+ mode will erase all data once the file is opened but after that we can do both read and write 

```python
f=open('Text.txt','a')
f.seek(0)
f.write("\nNew data")
f.close()
```

```python
f=open('Text.txt','a+')
f.seek(0)
f.write("\nNew data")
f.close()
```

```python
f=open('Text.txt')
print(f.read(5))
print(f.read())
print(f.read(7))
print(f.read())
f.close()
```

```python
f=open('Text.txt')
print(f.readline(3)) #Works same as f.read(3)
print(f.readline())
print(f.readline())
f.close()
```

```python
f=open('Text.txt')
x=f.readlines()
print(x)
print(x[3])
```
