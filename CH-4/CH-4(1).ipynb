```python
# write a python progrMm to count all the latters digites and spacial string from a given string
s="Py$@th%$13on"
a,b,c,d,e=0,0,0,0,0
for i in s:
    if i.isupper():
        a+=1
    elif i.islower():
        b+=1
    elif i.isdigit():
        c+=1
    elif i.isspace():
        d+=1
    else:
        e+=1
print("Uppercase:",a)
print("lowercase:",b)
print("digit:",c)
print("specialxase:",d)
print("space:",e)
```

```python
s="Py$@th54%$173on389"
a=0
count=0
for i in s:
    if i.isdigit():
        a += int(i)
        count+=1
print("sum:",a)
print("avarage:",a/count)
```
