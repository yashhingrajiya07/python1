```python
#WAP to ceheck validity of a password
#1. minimum 8 charcter
#2. alphabet must be [a-z]
#3. atleast one alphabet should be uppercase[A-Z]
#4. at least one digit
#5. atleast one character from [ _ OR $ OR @ ]

pas=(input("enter your password:"))
small=0
upper=0
digit=0
spe=0
if len(pas)<8:
    print("your password not valid")
else:
    for i in pas:
        if i.isupper():
            upper+=1
        if i.islower():
            small+=1
        if i.isdigit():
            digit+=1
        if i=='@' or i=='$' or i=='_':
            spe+=1
    if (small!=0 or upper!=0 or digit!=0 or spe!=0) and (len(pas)==(small+upper+digit+spe)) :
        print("Valid Password")
    else:
        print("Invalid Password")
            
```
