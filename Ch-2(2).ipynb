```python
#WAP to given num is happy num or not
n=int(input("enter number:"))
num=n
flag=False
sum=0
while True:
    while (n!=0):
        dig=n%10
        sum+=dig**2
        n//=10
    if sum==1:
        flag=True
        break
    elif sum==4:
        flag=False
        break
    else:
        n=sum
        sum=0
if flag==True:
    print(num,"is a happy number")
else:
    print(num,"is not a happy number")
```
