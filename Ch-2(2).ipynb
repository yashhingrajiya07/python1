## Iterative Statements
1. for loop
    Syntax:-
       for i in sequence:
               body of loop
           OR
       for i in range(5):
               print(i)

```python
for i in range(0,5):
    print("Hello")
    if i==3:
        print("Welcome")
```

```python
l=[1,2,6,7,8]
for i in l:
    print(i)
```

```python
i=1
while(i<=4):
    print(i)
    i=i+1
```

```python
# Nested Loop
for i in range(2):
    for j in range(2): # when i is 0, j will run twice again when i is 1, j will run twice.
        print(i,j)
```

```python
x=int(input("Enter the number of characters of your name:"))
name1=""
for i in range(x):
    name=input("Enter your name character:")
    name1+=name
print(name1)
```

```python
n=[1,2,3,4,5]
for i in n:
    print(i)
else:
    print("Loop completed without finding a specific condition")
```

```python
n=[1,2,3,4,5]
for i in n:
    if i==3:
        print("found 3, break the loop")
        break
    print(i)
else:
    print("Loop completed without finding a specific condition")
```

## break,continue,pass

```python
for i in range(10):
    if i==7:
        print("Process is enough, Now stop")
        break
        print(i)
```

```python
# continue:- to skip the iteration
for i in range(1,10):
    if i%2==0:
        continue
        print(i)
```

```python
# wap to check if a given year is leap year or not.\
year=int(input("Enter the year."))
if(year%4==0 and year%100!=0 or year%400==0):
    print("This is a leap year")
else:
    print("This is not a leap year")
```
